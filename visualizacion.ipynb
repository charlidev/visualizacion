```python
#importando matplotlib, pandas y seaborn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#cargando dataset desde seaborn
df = sns.load_dataset("iris")
df.head()
df.info()
```

```python
#creando la primera grafica con matplotlib
df["sepal_length"].plot()

#editando el tamaño de la grafica y el grosor de la linea y titulo
df["sepal_length"].plot(
    figsize=(10,5),
    linewidth=2,
    title="largo del sepalo",
    xlabel="indice de la observacion",
    ylabel="cantidad de cm",
    marker="o",
    linestyle="-."
)
```

```python
#hay otra forma de graficar y es pasando los arreglos a la funcion plot
plt.figure(figsize=(5,5))
plt.plot(df.index, df["sepal_length"])
plt.title("LArgo del sepalo")
plt.xlabel("Indice de la observacion")
plt.ylabel("Tamaño en cm")
plt.show()
```

```python
#Se pueden comparar varias cosas en una sola grafica
plt.figure(figsize=(10,5))
plt.plot(df.index, df["sepal_length"])
plt.plot(df.index, df["sepal_width"], color = "green", linewidth= 1, linestyle="--")
plt.title("LArgo del sepalo")
plt.xlabel("Indice de la observacion")
plt.ylabel("Tamaño en cm")
plt.legend(["largo", "ancho"])
plt.show()
```

SUBPLOTS

```python
#haciendo que tres graficas quepan en una sola fila

plt.figure(figsize=(15,4))

#inicia la personalizacion de la primera grafica
plt.subplot(1,3,1)
plt.plot(df.index,df["sepal_length"], color="green")
plt.title("LArgo del sepalo")
plt.xlabel("indice de observacion", loc="center")
plt.ylabel("tamaño en cm", loc="center")

#aqui empieza la personalizacion de la segunda grafica
plt.subplot(1,3,2)
plt.plot(df.index,df["sepal_width"], color="blue")
plt.title("Ancho del sepalo")
plt.xlabel("indice de observacion", loc="center")


#aqui empieza la personalizacion de la segunda grafica
plt.subplot(1,3,3)
plt.plot(df.index,df["petal_length"], color="red")
plt.title("largo  del petalo")
plt.xlabel("indice de observacion", loc="center")
```

```python
fig, axes = plt.subplots(1,3, figsize=(15,5), sharex=True, sharey=True)

#grafica 1
axes[0].plot(df.index, df["sepal_length"], color="green")
axes[0].set_title("Largo del sepalo")

#grafica 2
axes[1].plot(df.index, df["sepal_width"], color="blue")
axes[1].set_title("Largo del sepalo")

#grafica 3
axes[2].plot(df.index, df["petal_length"], color="red")
axes[2].set_title("Largo del sepalo")

fig.text(0.5, 0.03, "Indice de la observacion", ha="center", fontsize=12)
fig.text(0.04, 0.5, "tamaño en cm", va="center", fontsize=12, rotation="vertical")

plt.tight_layout(rect=[0.05, 0.05, 1,1])
plt.show()
```

```python
#creando un canvas vacio

fig = plt.figure()
ax1 = fig.add_axes([0.0, 0.0, 1.5, 1])
ax2 = fig.add_axes([0.1, 0.7, 0.3, 0.2])

#figura principal
ax1.plot(df.index, df["sepal_length"], "b")
ax1.set_xlabel("indice de la observacion")
ax1.set_ylabel("tamaño en cm")
ax1.set_title("largo de sepalo")

#figura principal
ax2.plot(df.index, df["sepal_width"], "r")
ax2.set_xlabel("indice de la observacion", fontdict={"size":9})
ax2.set_ylabel("tamaño en cm", fontdict={"size":9})
ax2.set_title("ancho de sepalo")
ax2.set_facecolor("lightgray")
```

GRAFICO DE BARRAS CON SEABORN

GRAFICOS DE BARRAS

```python
df["species"].value_counts().plot(
    kind="bar",
    title="Conteo de especies",
    color="green"
);
```

```python
#metodo funcional
plt.figure(figsize=(15,5))
plt.bar(x=df["species"].unique(), height=df["species"].value_counts())
plt.title("Conteo  x Especie")
plt.xlabel("Especie")
plt.ylabel("Conteo")
```

```python
#barras horizontales
plt.figure(figsize=(15,5))
plt.barh(y=df["species"].value_counts().index, width=df["species"].value_counts().values)
plt.title("Conteo  x Especie")
plt.xlabel("Especie")
plt.ylabel("Conteo")
```

```python
df_agrupado = df.groupby("species").agg({"sepal_length": "median"})
df_agrupado
```

```python
sns.barplot(data=df_agrupado, x=df_agrupado.index, y="sepal_length", hue="species")
plt.title("Barras verticales")
```

```python
sns.barplot(data=df_agrupado, y=df_agrupado.index, x="sepal_length", hue="species")
plt.title("Barras horizontales")
```

GRAFICAS DE PASTEL

```python
df_conteo_especies = df["species"].value_counts()
#el explode se usa para sobresaltar una sección de la grafica
df_conteo_especies.plot(kind="pie", figsize=(8,8), explode=[0.2,0,0], autopct="%1.1f%%")
```

```python
plt.figure(figsize=(8,8))
plt.pie(df["species"].value_counts(), explode=[0.2,0,0], autopct="%1.1f%%")
plt.title("distribucion de especies")
plt.legend(df["species"].value_counts().index)
plt.show()
```

boxplot

```python
colors = {
    "boxes": "DarkOrange",
    "whiskers": "Teal",
    "medians": "DarkBlue",
    "caps": "Gray"
}

df["sepal_length"].plot(
    kind="box",
    figsize=(6,6),
    patch_artist = True,
    showmeans = True,
    color = colors
)
```

```python
#en horizontal

plt.boxplot(df["sepal_length"], vert=False)
plt.show()
```

```python
#con seaborn
plt.figure(figsize=(15,5))
sns.boxplot(data=df, x="sepal_length", y="species", legend="full")
plt.show()
```

```python
ax = sns.boxplot(
    data=df,
    x="species",
    y="sepal_length",
    hue="species",
    palette="rainbow"
)

ax.set_title("largo del sepalo por especie")
ax.set_xlabel("especie")
ax.set_ylabel("tamaño en cm")

plt.show()
```
